```python
import sklearn
from sklearn.preprocessing import RobustScaler
```

```python
X=df1.drop(["y"],axis=1)
Y=df1["y"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,random_state=42,stratify=Y)
```

```python
from sklearn.preprocessing import LabelEncoder

label = LabelEncoder()

y_train = label.fit_transform(y_train)
y_test = label.transform(y_test)
```

```python
scaler = RobustScaler()
# Fit and scale training data
scaled_features=['x.radius_mean','x.texture_mean','x.area_se','x.texture_worst','x.area_mean','x.radius_worst','x.perimeter_mean','x.perimeter_worst','x.area_worst']
x_train[scaled_features] = scaler.fit_transform(
    x_train[scaled_features]
)

# Scale testing data using the same scaler
x_test[scaled_features] = scaler.transform(
    x_test[scaled_features]
)
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score,confusion_matrix,ConfusionMatrixDisplay,precision_score,recall_score,f1_score
```

```python
model=SVC(kernel='linear')
model.fit(x_train,y_train)
y_pred1=model.predict(x_train)
print("Precision:",precision_score(y_train,y_pred1))
print("Recall:",recall_score(y_train,y_pred1))
print("F1-score:",f1_score(y_train,y_pred1))
print("Accuracy:",accuracy_score(y_train,y_pred1))
```

```python
y_pred2=model.predict(x_test)
```

```python
print("Precision:",precision_score(y_test,y_pred2))
print("Recall:",recall_score(y_test,y_pred2))
print("F1-score:",f1_score(y_test,y_pred2))
print("Accuracy:",accuracy_score(y_test,y_pred2))
```

```python
cm=confusion_matrix(y_test,y_pred2)
```

```python
display=ConfusionMatrixDisplay(confusion_matrix=cm)
display.plot()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import RobustScaler
from sklearn.svm import SVC

# Select only two features for visualization
features = ['x.radius_mean', 'x.texture_mean']

X_two = x_train[features]
y = y_train

# Train a separate 2-feature SVC model
model_2d = make_pipeline(
    RobustScaler(),
    SVC(kernel='linear')
)
model_2d.fit(X_two, y)

# Create a grid for plotting
x_min, x_max = X_two.iloc[:, 0].min() - 1, X_two.iloc[:, 0].max() + 1
y_min, y_max = X_two.iloc[:, 1].min() - 1, X_two.iloc[:, 1].max() + 1

xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 300),
    np.linspace(y_min, y_max, 300)
)

grid = np.c_[xx.ravel(), yy.ravel()]
Z = model_2d.predict(grid).reshape(xx.shape)

# Plot the decision boundary and training points
plt.contourf(xx, yy, Z, alpha=0.25, cmap='coolwarm')
plt.scatter(
    X_two.iloc[:, 0],
    X_two.iloc[:, 1],
    c=y,
    cmap='coolwarm',
    edgecolors='black'
)

plt.xlabel(features[0])
plt.ylabel(features[1])
plt.title('SVC Decision Boundary (Linear Kernel, 2 Features)')
plt.show()
```
